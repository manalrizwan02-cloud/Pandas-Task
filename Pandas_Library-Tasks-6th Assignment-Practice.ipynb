```python
#Task 1
#Loc
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Sana", "Hina"],
    "Age": [20, 21, None, 22, 19],
    "Marks": [75, None, 65, 92, 80],
    "City": ["Karachi", "Lahore", "Islamabad", None, "Karachi"]
}
df=pd.DataFrame(data)
print(df)
print("\n",df.loc[2])
print("\n",df.loc[1,["Name","Marks"]])
print("\n",df.loc[[0,3]])
print("\n",df.loc[[0,3],["Name","City"]])
```

```python
#task 2
#Iloc
data = {

    "Name": ["Ali", "Sara", "Ahmed", "Sana", "Hina"],
    "Age": [20, 21, None, 22, 19],
    "Marks": [75, None, 65, 92, 80],
    "City": ["Karachi", "Lahore", "Islamabad", None, "Karachi"]

}
df =pd.DataFrame(data)
print(df)

print("\n",df.iloc[2])
print("\n",df.iloc[[0,2]])
print("\n",df.iloc[[0,2],[0,2]])
print("\n",df.iloc[[3,1]])
```

```python
#task 3
#Deleting raw
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Sana", "Hina"],
    "Age": [20, 21, None, 22, 19],
    "Marks": [75, None, 65, 92, 80],
    "City": ["Karachi", "Lahore", "Islamabad", None, "Karachi"]

}
df =pd.DataFrame(data)
print("\n",df)
df =df.drop(1,axis=0 )
print("\n",df)
df=df.drop([0,4],axis=0)
print("\n",df)
```

```python
#Task 4
#Delete columns
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Sana", "Hina"],
    "Age": [20, 21, None, 22, 19],
    "Marks": [75, None, 65, 92, 80],
    "City": ["Karachi", "Lahore", "Islamabad", None, "Karachi"]

}
df =pd.DataFrame(data)
print("\n",df)
df=df.drop("Age",axis=1)
print("\n",df)
df=df.drop(["Marks","City"],axis=1)
print("\n",df)
```

```python
#Task5
#Missing Values
data = {

    "Name": ["Ali", "Sara", "Ahmed", "Sana", "Hina"],
    "Age": [20, 21, None, 22, 19],
    "Marks": [75, None, 65, 92, 80],
    "City": ["Karachi", "Lahore", "Islamabad", None, "Karachi"]

}
df =pd.DataFrame(data)
print("\n",df)

print("\n",df.isnull())

print("\n",df.isnull().sum())

df["Age"]=df["Age"].fillna(0)
print("\n",df)

df["City"]=df["City"].fillna("Unknown")
print("\n",df)

df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
print("\n",df)

print("\n",df.dropna())
```
